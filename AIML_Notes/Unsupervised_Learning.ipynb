```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
uploaded = files.upload()

df = pd.read_csv("income.csv")
print(df.head())
```

**KMeans**

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder

data = df.copy()

for col in data.columns:
    if data[col].dtype == 'object':
        data[col] = LabelEncoder().fit_transform(data[col].astype(str))

data = data.select_dtypes(include=['int64', 'float64'])

kmeans = KMeans(n_clusters=3, random_state=42)
data["Cluster"] = kmeans.fit_predict(data)

print(data)

plt.scatter(data.iloc[:,0], data.iloc[:,1], c=data["Cluster"], s=100)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1],
            color='red', marker='X', s=250)
plt.xlabel(data.columns[0])
plt.ylabel(data.columns[1])
plt.title("K-Means Clustering")
plt.show()
```

**DBSCAN**

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler

db_data = data.drop("Cluster", axis=1)

scaled = StandardScaler().fit_transform(db_data)

dbscan = DBSCAN(eps=0.8, min_samples=5)

db_data["Cluster"] = dbscan.fit_predict(scaled)

print(db_data)

plt.scatter(db_data. Iniloc[:,0], db_data.iloc[:,1],
            c=db_data["Cluster"], s=100)
plt.xlabel(db_data.columns[0])
plt.ylabel(db_data.columns[1])
plt.title("DBSCAN Clustering")
plt.show()
```

**Aprior**

```python
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd

onehot = pd.get_dummies(df.astype(str))

frequent_itemsets = apriori(onehot, min_support=0.05, use_colnames=True)
print(frequent_itemsets)

rules = association_rules(frequent_itemsets,
                          metric="confidence",
                          min_threshold=0.6)

print(rules[['antecedents','consequents','support','confidence','lift']])

if not frequent_itemsets.empty:
    plt.bar(range(len(frequent_itemsets)), frequent_itemsets['support'])
    plt.xticks(range(len(frequent_itemsets)),
               frequent_itemsets['itemsets'],
               rotation=90)
    plt.title("Apriori Frequent Itemsets")
    plt.ylabel("Support")
    plt.show()
else:
    print("No frequent itemsets found to plot.")
```

**PCA**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

data = {
    'Age':[25,35,45,55,65],
    'BP':[120,130,140,150,160],
    'Cholesterol':[180,200,220,240,260],
    'Sugar':[90,100,120,140,160],
    'BMI':[22,24,27,30,33]
}

df = pd.DataFrame(data)

pca = PCA(n_components=2)
reduced = pca.fit_transform(df)

plt.figure(figsize=(6,4))
plt.scatter(reduced[:,0], reduced[:,1])

for i, (x, y) in enumerate(reduced):
    plt.annotate(f'Sample {i+1}', (x, y))

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Visualization')
plt.grid(True)
plt.show()
```
