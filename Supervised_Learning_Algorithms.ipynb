**Simple Linear Regression**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5, 6],
    'Marks': [35, 40, 50, 55, 65, 70]
})

X = df[['Hours']]
y = df['Marks']

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[7]])
print(prediction[0])
pred = model.predict(X)

plt.scatter(df['Hours'], df['Marks'], label='Actual Data')
plt.plot(df['Hours'], pred, color='red', label='Regression Line')
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

**Multiple Linear Regression**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5],
    'Attendance': [60, 65, 70, 80, 90],
    'Marks': [30, 40, 50, 60, 75]
})

X = df[['Hours', 'Attendance']]
y = df['Marks']

model = LinearRegression()
model.fit(X, y)
prediction = model.predict([[6,95]])
print(prediction[0])
pred = model.predict(X)

plt.scatter(df['Hours'], df['Marks'], label='Actual Data')
plt.plot(df['Hours'], pred, color='red', label='Regression Line')
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Multiple Linear Regression")
plt.legend()
plt.show()
```

**Decision Tree**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

df = pd.DataFrame({
    'Experience': [1,2,3,4,5,6,7,8],
    'Salary': [25000,30000,40000,45000,50000,60000,65000,70000]
})

X = df[['Experience']]
y = df['Salary']

model = DecisionTreeRegressor(random_state=0)
model.fit(X, y)
prediction = model.predict([[10]])
print(prediction[0])
pred = model.predict(X)

plt.scatter(df['Experience'], df['Salary'])
plt.plot(df['Experience'], pred, color='red')
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Decision Tree Regression")
plt.show()
```

**Logistic Regression**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

df = pd.DataFrame({
    'Age': [12, 17, 18, 23, 35, 40],
    'Licence': [0, 0, 1, 1, 1, 1]
})

X = df[['Age']]
y = df['Licence']

model = LogisticRegression()
model.fit(X, y)

pred = model.predict(X)

prediction = model.predict([[50]])
if prediction[0] == 1:
    print("Age 50: Eligible for Licence")
else:
    print("Age 50: Not Eligible for Licence")

plt.scatter(df['Age'], df['Licence'], label='Actual Data')
plt.plot(df['Age'], pred, color='red', label='Prediction')
plt.xlabel("Age")
plt.ylabel("Licence (0/1)")
plt.title("Logistic Regression Classification")
plt.legend()
plt.show()
```

**Navie Bayes Classification**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB

df = pd.DataFrame({
    'Age': [18, 22, 25, 30, 35, 40],
    'Buy': [0, 0, 0, 1, 1, 1]
})

X = df[['Age']]
y = df['Buy']

model = GaussianNB()
model.fit(X, y)

pred = model.predict(X)

# Check for age 50
prediction = model.predict([[50]])
if prediction[0] == 1:
    print("Age 50: Will Buy")
else:
    print("Age 50: Will Not Buy")

plt.scatter(df['Age'], df['Buy'], label='Actual Data')
plt.plot(df['Age'], pred, color='red', label='Prediction')
plt.xlabel("Age")
plt.ylabel("Buy (0/1)")
plt.title("Naive Bayes Classification")
plt.legend()
plt.show()
```

**Support Vector Machine**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.svm import SVC

df = pd.DataFrame({
    'Marks': [20, 30, 40, 50, 60, 70],
    'Pass':  [0, 0, 0, 1, 1, 1]
})

X = df[['Marks']]
y = df['Pass']

model = SVC(kernel='linear')
model.fit(X, y)

prediction = model.predict([[55]])
print("Prediction:", prediction[0])

pred = model.predict(X)

plt.scatter(df['Marks'], df['Pass'], label='Actual Data')
plt.plot(df['Marks'], pred, color='red', label='Prediction')
plt.xlabel("Marks")
plt.ylabel("Pass (0/1)")
plt.title("Support Vector Machine")
plt.legend()
plt.show()
```

**Polynomial Regression**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    'Hours': [1,2,3,4,5,6],
    'Marks': [30,35,45,60,80,105]
})

X = df[['Hours']]
y = df['Marks']

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

prediction = model.predict(poly.transform([[7]]))
print("Predicted Marks:", prediction[0])

pred = model.predict(X_poly)

plt.scatter(df['Hours'], df['Marks'], label='Actual Data')
plt.plot(df['Hours'], pred, color='red', label='Polynomial Curve')
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("Polynomial Regression")
plt.legend()
plt.show()
```

**KNN**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

df = pd.DataFrame({
    'Height': [150,155,160,165,170,175],
    'Sports': [0,0,0,1,1,1]
})

X = df[['Height']]
y = df['Sports']

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

prediction = model.predict([[168]])
print("Prediction:", prediction[0])

pred = model.predict(X)

plt.scatter(df['Height'], df['Sports'], label='Actual Data')
plt.plot(df['Height'], pred, color='red', label='Prediction')
plt.xlabel("Height")
plt.ylabel("Sports (0/1)")
plt.title("K-Nearest Neighbors")
plt.legend()
plt.show()
```

**Random Forest**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier

df = pd.DataFrame({
    'Temperature': [20,22,25,28,30,35],
    'Play': [0,0,1,1,1,0]
})

X = df[['Temperature']]
y = df['Play']

model = RandomForestClassifier(n_estimators=10, random_state=42)
model.fit(X, y)

prediction = model.predict([[27]])
print("Prediction:", prediction[0])

pred = model.predict(X)

plt.scatter(df['Temperature'], df['Play'], label='Actual Data')
plt.plot(df['Temperature'], pred, color='red', label='Prediction')
plt.xlabel("Temperature")
plt.ylabel("Play (0/1)")
plt.title("Random Forest")
plt.legend()
plt.show()
```
